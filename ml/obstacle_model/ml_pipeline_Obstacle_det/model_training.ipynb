```python
import os
import shutil
import random

# Use 'r' before the quotes to treat backslashes as literal text
source_img = r'C:\Users\USER\OneDrive - Siri Sunanda Dhamma School\Desktop\Obstacle_Detection_Project\drishti-final\ml_pipeline_Obstacle_det\drishti_final_dataset\images'
source_lbl = r'C:\Users\USER\OneDrive - Siri Sunanda Dhamma School\Desktop\Obstacle_Detection_Project\drishti-final\ml_pipeline_Obstacle_det\drishti_final_dataset\labels'
dest_base = r'C:\Users\USER\OneDrive - Siri Sunanda Dhamma School\Desktop\Obstacle_Detection_Project\drishti-final\ml_pipeline_Obstacle_det\drishti_final_split'

# Create the folder structure
for split in ['train', 'val', 'test']:
    os.makedirs(os.path.join(dest_base, 'images', split), exist_ok=True)
    os.makedirs(os.path.join(dest_base, 'labels', split), exist_ok=True)

# This part will now work because source_img is correctly identified
files = [f[:-4] for f in os.listdir(source_img) if f.endswith('.jpg')]
random.shuffle(files)

# Calculate split points for your 6.4k images
train_end = int(len(files) * 0.7)
val_end = int(len(files) * 0.9)

# Move files using f-strings with correct path joining
for i, name in enumerate(files):
    split = 'train' if i < train_end else 'val' if i < val_end else 'test'
    shutil.copy(os.path.join(source_img, f"{name}.jpg"), os.path.join(dest_base, 'images', split, f"{name}.jpg"))
    shutil.copy(os.path.join(source_lbl, f"{name}.txt"), os.path.join(dest_base, 'labels', split, f"{name}.txt"))

print(f"Dataset split complete in {dest_base}!")
```
