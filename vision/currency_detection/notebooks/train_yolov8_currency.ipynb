```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8m.pt")
```

```python
model.train(
    data="/kaggle/input/sl-currency-dataset-yolov/sl-currency-dataset-yolov/dataset.yaml",
    epochs=50,
    patience=20,
    imgsz=640,
    batch=16,
    device=0,

    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=10,
    translate=0.1,
    scale=0.5,
    shear=2,
    perspective=0.0005,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.2
)
```

```python
metrics = model.val()
print(metrics)
```

```python
model.export(format="onnx")
```
