```python
import cv2
import win32com.client
import time
import threading
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
from ultralytics import YOLO

# 1. Setup Audio (Windows built-in)
speaker = win32com.client.Dispatch("SAPI.SpVoice")

def speak(text):
    def run(): 
        speaker.Speak(text)
    threading.Thread(target=run).start()

# 2. Load model from your local Downloads path
model_path = r'.\ml_pipeline_Obstacle_det\runs\detect\Drishti_Final_Push\v11n_augmented_852\weights\best.pt'
model = YOLO(model_path)

# 3. Open Video File from your 'vedios' folder
video_file = r'vedios\test1.mp4' 
cap = cv2.VideoCapture(video_file)

# Dictionary to track announcement times for the 3-second cooldown
last_announced = {}

print(f"Processing Video: {video_file}. Press stop in the notebook to quit.")

try:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret: 
            break 

        # 4. Run Inference (0.5 confidence for safety)
        results = model.predict(frame, conf=0.5, verbose=False)

        # 5. Voice Logic with Cooldown
        detected_classes = [model.names[int(box.cls[0])] for box in results[0].boxes]
        
        for class_name in set(detected_classes):
            now = time.time()
            if class_name not in last_announced or (now - last_announced[class_name]) > 3:
                speak(f"Object ahead: {class_name}")
                last_announced[class_name] = now

        # 6. FIXED: Notebook-Compatible Display using Matplotlib
        annotated_frame = results[0].plot()
        frame_rgb = cv2.cvtColor(annotated_frame, cv2.COLOR_BGR2RGB)
        
        plt.figure(figsize=(10, 6))
        plt.imshow(frame_rgb)
        plt.axis('off')
        display(plt.gcf())
        clear_output(wait=True)
        plt.close()

finally:
    # Release the video file correctly
    cap.release()
    print("Video processing completed.")
```
